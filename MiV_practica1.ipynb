# Imports

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

# Revisió de dades

```python
df = pd.read_csv('original_data.csv', sep=";")
df.info()
```

# Processament de dades

```python
df["Population"] = df["Population"].apply(lambda x: int(x.replace(',','')) if x != "--" else None)
df["Total Murders"] = df["Total Murders"].apply(lambda x: int(x.replace(',','')) if x != "--" else None)
df["Murders Rate per 100,000"] = df["Murders Rate per 100,000"].apply(lambda x: float(x) if x != "--" else None)
df["Total by Firearm"] = df["Total by Firearm"].apply(lambda x: int(x) if x != "--" else None)
df["Total Handgun"] = df["Total Handgun"].apply(lambda x: int(x) if x != "--" else None)
df["Total Other Firearm"] = df["Total Other Firearm"].apply(lambda x: int(x) if x != "--" else None)
df["Total Percent by Firearms"] = df["Total Percent by Firearms"].apply(lambda x: float(x) if x != "--" else None)
df["Percent by Handguns"] = df["Percent by Handguns"].apply(lambda x: float(x) if x != "--" else None)
df["Percent by Other Firearms"] = df["Percent by Other Firearms"].apply(lambda x: float(x) if x != "--" else None)
df["Total by Firearm Percentage change"] = df["Total by Firearm Percentage change"].apply(lambda x: float(x) if x != "--" else None)
df["Population Percentage Change"] = df["Population Percentage Change"].apply(lambda x: float(x) if x != "--" else None)
```

```python
df.info()
```

# Normalització de les dades

```python
df["Murder Rate by Firearm per 100000"] = ((df["Total by Firearm"] * 100000) / df["Population"]).round(1)
df.info()
df.to_csv("Florida_Gun_Death_Rate.csv", index=False)
```

```python
df = pd.read_csv("Florida_Gun_Death_Rate.csv")
df = df.iloc[19:]
```

# Representació de Gràfics

### Murder Rate Graph

```python
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df["Year"], y=df["Total Murders"],
    mode='lines',
    line=dict(width=2, color='rgba(160, 160, 160, 0.8)', shape='spline'),
    fill='tozeroy',
    fillcolor='rgba(160, 160, 160, 0.1)',
    name="Total Murders",
    hovertemplate="Total: %{y}<extra></extra>"
))
fig.add_trace(go.Scatter(
    x=df["Year"], y=df["Total by Firearm"],
    mode='lines+markers',
    line=dict(width=3, color='rgb(220, 20, 60)', shape='spline'),
    marker=dict(size=7, color='red', line=dict(width=2, color='rgb(220, 20, 60)')),
    fill='tozeroy',
    fillcolor='rgba(220, 20, 60, 0.15)',
    name="Murder by Firearm",
    hovertemplate="By Firearm: %{y}<extra></extra>"
))

fig.update_layout(
    title=dict(text="Murder Rate by Firearm in Florida", font=dict(size=22)),
    xaxis_title="Year", yaxis_title="Number of Murders", template="plotly_white",
    legend=dict(bgcolor="rgba(255, 255, 255, 0.8)"),
    margin=dict(l=60, r=40, t=80, b=50)
)
fig.update_xaxes(dtick=1, showgrid=False, tickangle=45)
fig.update_yaxes(showgrid=True, gridcolor='rgba(200, 200, 200, 0.3)', zeroline=False)

fig.add_annotation(
    x=2005, y=521, 
    text="<b>2005:</b><br>Florida enacted<br>'Stand your Ground'",
    showarrow=True,
    arrowhead=2,
    arrowcolor="black",
    arrowsize=1,
    arrowwidth=2,
    ax=80, ay=50,
    bordercolor="black",
    borderwidth=1.5,
    borderpad=6,
    bgcolor="white",
    font=dict(size=12, color="black"),
    align="left"
)
fig.show()
```
